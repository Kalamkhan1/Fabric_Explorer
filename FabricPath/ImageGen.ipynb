```python
# Import libraries
from flask import Flask, request, send_file
from diffusers import StableDiffusionPipeline
import torch
from pyngrok import ngrok
from diffusers import DiffusionPipeline


# Initialize the Stable Diffusion pipeline
print("Loading Stable Diffusion model...")
pipe = DiffusionPipeline.from_pretrained("stabilityai/stable-diffusion-3.5-medium")
pipe.to("cuda")  # Use GPU

# Set up Flask app
app = Flask(__name__)

@app.route("/generate", methods=["GET"])
def generate_image():
    # Get the prompt from the query parameter
    prompt = request.args.get("prompt", default="A beautiful sunset over the mountains")

    # Generate the image
    print(f"Generating image for prompt: {prompt}")
    image = pipe(prompt).images[0]
    output_path = "/content/generated_image.png"
    image.save(output_path)

    # Serve the generated image file
    return send_file(output_path, mimetype="image/png")

# Set up ngrok
public_url = ngrok.connect(5000)
print(f"Access the API at: {public_url}/generate")

# Run the Flask app
app.run(host="0.0.0.0", port=5000)
```
